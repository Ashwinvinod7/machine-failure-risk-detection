### STATISTICAL ANAMOLY DETECTION and preprocessing

```python
df1=df[SENSOR_COLS]
```

```python
df1
```

#### z-score

```python
z = np.abs(stats.zscore(df1))
flag_z = (z > 3).any(axis=1)
```

```python
z
```

```python
flag_z
```

```python
q1=df1.quantile(0.25)
q3=df1.quantile(0.75)
```

```python
q1
```

```python
q3
```

```python
IQR = q3-q1
```

```python
IQR
```

```python
minimum=q1-(1.5*IQR)
maximum=q3+(1.5*IQR)
```

```python
minimum
```

```python
maximum
```

```python
print(f'minimum IQR value  is {minimum}')
print(f'maximum IQR value  is {maximum}')
```

```python

flag_IQR = ((df1 < q1 - 1.5 * IQR) | (df1 > q3 + 1.5 * IQR)).any(axis=1)
```

```python
flag_IQR
```

#### mahanobolis threshold

```python
diff = df1.values - df1.values.mean(axis=0)
inv_cov = np.linalg.pinv(np.cov(df1.values, rowvar=False))
mahalanobis = np.sqrt(np.einsum("ij,jk,ik->i", diff, inv_cov, diff))
flag_mahalanobis = mahanobolis > np.sqrt(stats.chi2.ppf(0.999, df=len(SENSOR_COLS)))
```

```python
diff
```

```python
inv_cov 
```

```python
mahalanobis
```

```python
flag_mahalanobis
```

#### isolation forest

```python
from sklearn.ensemble import IsolationForest
isolation = IsolationForest(contamination=0.03, random_state=42).fit(df1)
flag_iso = isolation.predict(df1) == -1
```

```python
isolation
```

```python
flag_iso
```

```python
anom = pd.DataFrame({"Z-score": flag_z, "IQR": flag_IQR,
                     "Mahalanobis": flag_mahalanobis, "IsolationForest": flag_iso})
anom["votes"] = anom.sum(axis=1)
```

```python
anom
```

```python
anom.drop(columns="votes").sum()
```

```python
overall = pd.DataFrame({
    "flagged": anom.drop(columns="votes").sum(),
    "% of data": anom.drop(columns="votes").mean().mul(100).round(2),
    "failure rate among flagged (%)": [df.loc[anom[c], "failure"].mean() * 100
                                       for c in anom.columns[:-1]],
})
```
